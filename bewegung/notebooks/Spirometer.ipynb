# Spirometer Parameter

Column 10: time  
Column 11: Rf [1/min] Respiratorische Frequenz in Atemzügen pro Minute (Atemfrequenz)  
12: VT [L] Atemzugvolumen  
13: VE [L/min] Atemminutenvolumen  
14: IV [mL] Inspiratorische Vitalkapazität (Luftvolumen, das man max. einatmen kann)  
15: VO2 [mL/min] maximale Sauerstoffaufnahme (Indikator für Ausdauerleistungsfähigkeit)  
16: VCO2 [ml/min] Ausatmungs-Kohlendioxid-Volumen  
17: RQ=[VCO2]/[VO2] Respiratorischer Quotient  
18: VE/VO2  
19: VE/VCO2  
20: VO2/kg [mL/min/Kg]  
21: METS  
22: FeO2 [%]  
23: FeCO2 [%]  
24: FiO2 [%]  
25: FiCO2 [%]  
26: PeO2 [mmHg]  
27: PeCO2 [mmHg]

```python
# Pfad zur Datei
spiro_1_P03 = pd.read_excel("C:/Users/malin/HESSENBOX-DA/AnalyseSyntheseMenschlicherBewegung/Metabolic data (Cosmed K5)/Metabolic data (Cosmed K5)/P03 P03 20251113 CPET MIX_20251124134235.xlsx")
spiro_2_P03 = pd.read_excel("C:/Users/malin/HESSENBOX-DA/AnalyseSyntheseMenschlicherBewegung/Metabolic data (Cosmed K5)/Metabolic data (Cosmed K5)/P03 P03 20251113 CPET MIX_20251124134259.xlsx")
spiro_3_P03 = pd.read_excel("C:/Users/malin/HESSENBOX-DA/AnalyseSyntheseMenschlicherBewegung/Metabolic data (Cosmed K5)/Metabolic data (Cosmed K5)/P03 P03 20251113 CPET MIX_20251124134315.xlsx")
spiro_4_P03 = pd.read_excel("C:/Users/malin/HESSENBOX-DA/AnalyseSyntheseMenschlicherBewegung/Metabolic data (Cosmed K5)/Metabolic data (Cosmed K5)/P03 P03 20251113 CPET MIX_20251124134338.xlsx")
# Speichert Versuchsperson und Uhrzeit
Person_1_P03 = spiro_1_P03.iloc[1,1]
Uhrzeit_1_P03 = spiro_1_P03.iloc[0,4]

Person_2_P03 = spiro_2_P03.iloc[1,1]
Uhrzeit_2_P03 = spiro_2_P03.iloc[0,4]

Person_3_P03 = spiro_3_P03.iloc[1,1]
Uhrzeit_3_P03 = spiro_3_P03.iloc[0,4]

Person_4_P03 = spiro_4_P03.iloc[1,1]
Uhrzeit_4_P03 = spiro_4_P03.iloc[0,4]
```

```python
# Zeit auslesen
def zeit_in_sekunden(z):
    if pd.isna(z):
        return np.nan

    # Excel-Zeit (datetime.time)
    if isinstance(z, dt.time):
        return z.hour * 3600 + z.minute * 60 + z.second

    # String
    z = str(z).strip()

    # reine Einheiten wie "s", "ms"
    if z.isalpha():
        return np.nan

    # gültige Zeitstrings wie "00:00:10"
    return pd.to_timedelta(z, errors="coerce").total_seconds()

spiro_1_P03["t_s"] = spiro_1_P03["t"].apply(zeit_in_sekunden)
spiro_2_P03["t_s"] = spiro_2_P03["t"].apply(zeit_in_sekunden)
spiro_3_P03["t_s"] = spiro_3_P03["t"].apply(zeit_in_sekunden)
spiro_4_P03["t_s"] = spiro_4_P03["t"].apply(zeit_in_sekunden)

# ungültige Zeilen entfernen
spiro_1_P03 = spiro_1_P03.dropna(subset=["t_s"])
spiro_2_P03 = spiro_2_P03.dropna(subset=["t_s"])
spiro_3_P03 = spiro_3_P03.dropna(subset=["t_s"])
spiro_4_P03 = spiro_4_P03.dropna(subset=["t_s"])

#print(spiro["t"])

#print(spiro["t"].head())
#print(spiro["t"].dtype)
```

```python
#print(spiro["VT"])
```

## Plot VO2

```python
plt.figure()
#plt.plot(spiro_1.loc[3:,"t_s"], spiro_1.loc[3:,"VT"])
datasets = [
    (spiro_1_P03,  "Asphalt"),
    (spiro_2_P03, "Gras"),
    (spiro_3_P03, "Sand"),
    (spiro_4_P03, "Wald"),
]

for df, label in datasets:
    plt.plot(
        df.loc[3:, "t_s"],
        df.loc[3:, "VO2"],
        label=label
    )
plt.xlabel("Zeit [s]")
plt.ylabel("VO2 [ml/min]")
plt.grid(True)
plt.title(Person_1_P03)
#plt.title(f"{Person_1} - {Uhrzeit_1}")
plt.legend()
plt.show()
```
