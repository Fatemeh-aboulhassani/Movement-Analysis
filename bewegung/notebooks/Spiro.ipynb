```python
import os
import math

import numpy as np
import pandas as pd

import scipy.stats as stats

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#path
path = r"C:\Users\abolhassni\Desktop\TU Darmstadt\Term 2\bewegung\Bewegunganalysieren\Bewegung\data\raw\Metabolic data (Cosmed K5)\Metabolic data (Cosmed K5) - Copy"

# read all data when that is .txt
file_names = [f for f in os.listdir(path) if f.endswith('.xlsx')]

#clear name and remove space and - from name
def sanitize_name(name: str) -> str:
    name = name.replace(".xlsx", "")
    name = name.replace(" ", "_").replace("-", "_")
    name = ''.join(c if c.isalnum() or c == '_' else '_' for c in name)
    if name and name[0].isdigit():
        name = '_' + name
    return name


list_name_dataframe=[]
# read all data
for file_name in file_names:
    var_name = sanitize_name(file_name)
    file_path = os.path.join(path, file_name)
    
    print(f" file: {var_name} -- name dataframe: {file_name}")
    
    try:
        df = pd.read_csv(
            file_path,
            sep="\t",
            decimal=",",
            engine="python"
        )
        df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
        
        #creat one liest by name of datafram
        list_name_dataframe.append(var_name)

        # name of data frame
        globals()[var_name] = df
        
        
    except Exception as e:
        print(f" --------Error---- {file_name}: {e}")

```
