```python
import math

import os
import shutil

import glob

import numpy as np
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt
```

```python
input_file=r"C:/Users/abolhassni/Desktop/kinematic/movement-analysis/data/raw/Kinematic data (xsens suit)"
output_file=r"C:/Users/abolhassni/Desktop/TU Darmstadt/Term 2/bewegung/Bewegunganalysieren/Bewegung/data/processed/person_id"
files=os.listdir(input_file)


for file in files:
    if file.endswith(".txt"):
        print(file)
        parts=file.split("_")
        person_part=parts[2]
        person_id=person_part.split("-")[0]
        person_folder=os.path.join(output_file,person_id)
        os.makedirs(person_folder, exist_ok=True)
        source = os.path.join(input_file, file)
        destination = os.path.join(person_folder, file)

        
        shutil.copy(source, destination)
        
```

```python

input_folder = r"C:\Users\abolhassni\Desktop\TU Darmstadt\Term 2\bewegung\Bewegunganalysieren\Bewegung\data\processed\person_id"
output_folder = r"C:\Users\abolhassni\Desktop\TU Darmstadt\Term 2\bewegung\Bewegunganalysieren\Bewegung\data\processed\person_noheader"

for folder in os.listdir(input_folder):
    input_folder_path = os.path.join(input_folder, folder)
    
    if os.path.isdir(input_folder_path):
        
    
        output_subfolder = os.path.join(output_folder, folder)
        os.makedirs(output_subfolder, exist_ok=True)
        
        for file in os.listdir(input_folder_path):
            if file.endswith(".txt"):
                
                input_file_path = os.path.join(input_folder_path, file)
                output_file_path = os.path.join(output_subfolder, file)
                
            
                with open(input_file_path, "r") as f:
                    lines = f.readlines()
                
            
                new_lines = lines[12:]
                
              
                with open(output_file_path, "w") as f:
                    f.writelines(new_lines)
                
                print(f"Saved: {output_file_path}")
```

```python


input_folder = r"C:\Users\abolhassni\Desktop\TU Darmstadt\Term 2\bewegung\Bewegunganalysieren\Bewegung\data\processed\person_noheader"
output_folder = r"C:\Users\abolhassni\Desktop\TU Darmstadt\Term 2\bewegung\Bewegunganalysieren\Bewegung\data\processed\person_quat"

for folder in os.listdir(input_folder):
    input_folder_path = os.path.join(input_folder, folder)
    
    if os.path.isdir(input_folder_path):
        
        output_subfolder = os.path.join(output_folder, folder)
        os.makedirs(output_subfolder, exist_ok=True)
        
        for file in os.listdir(input_folder_path):
            if file.endswith(".txt"):
                
                input_file_path = os.path.join(input_folder_path, file)
                output_file_path = os.path.join(output_subfolder, file)
                
                try:
                   
                    df = pd.read_csv(input_file_path, sep="\t")
                    
               
                    cols = ["Quat_q0", "Quat_q1", "Quat_q2", "Quat_q3"]
                    df_selected = df[cols]
                    
             
                    df_selected.to_csv(output_file_path, sep="\t", index=False)
                    
                    print(f"Saved: {output_file_path}")
                
                except Exception as e:
                    print(f"Error in {file}: {e}")
```
